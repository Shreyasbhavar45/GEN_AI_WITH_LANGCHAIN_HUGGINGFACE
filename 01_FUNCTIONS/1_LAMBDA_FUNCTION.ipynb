## __LAMBDA FUNCTIONS__
- Lambda functions are small anonymous functions defined using the `lambda` keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.

### Lambda function in simple language
- A lambda function is a small function without a name.
- It is used when we need a simple function for a short time.
- It is written in one line.
- It can take input values called arguments.
- It returns the result automatically.
- Syntax: `lambda arguments: expression`

```python
# Simple lambda example
square = lambda x: x * x
print(square(5))
```

```python
## syntax of Lambda function
lambda argument : expression
```

```python
#Example of lambda function.
def addition(a,b):
    return a+b

addition(2,3)
```

```python
#convert to lamda function

addition = lambda a,b:a+b
# type(addition)
print(addition(5,6))
```

```python
#example 2
def even(num):
    if num % 2 == 0:
        return True
    
even(24)
```

```python
#convert To lambda Function
even1 = lambda num:num%2==0
print(even1(76))
```

```python
def addition(x,y,z):
    return x+y+z

addition(12,3,5)
```

```python
#convert to lambda function
addition1 = lambda x,y,z : x + y + z
print(addition1(12,3,5))
```
