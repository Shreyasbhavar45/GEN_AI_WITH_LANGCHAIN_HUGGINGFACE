## __DIFFERENCE BETWEEN LAMBDA, MAP, AND FILTER__

### Simple meaning
- `lambda` is used to create a small function in one line.
- `map()` is used to apply a function to every item.
- `filter()` is used to keep only items that match a condition.

### Main difference in points

| Function | What it does | Output |
|---|---|---|
| `lambda` | Creates a small anonymous function | A function |
| `map()` | Applies a function to every item | All transformed items |
| `filter()` | Checks a condition for every item | Only selected items |

### Easy way to remember
- `lambda` means make a small function.
- `map()` means change every item.
- `filter()` means choose only some items.

### 1. Lambda function
- Lambda is a short way to write a function.
- It has no function name.
- It is useful for small logic.
- It can be used alone or inside `map()` and `filter()`.

```python
# Lambda example
square = lambda num: num * num

print(square(5))
```

### 2. Map function
- `map()` applies a function to every item.
- It is used when we want to transform all values.
- Number of items in output is usually same as input.
- Example: convert every number into its square.

```python
# Map example
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda num: num * num, numbers))

print(squares)
```

### 3. Filter function
- `filter()` checks every item using a condition.
- It keeps only items where the condition is `True`.
- It is used when we want to select values.
- Number of items in output can be less than input.
- Example: keep only even numbers.

```python
# Filter example
numbers = [1, 2, 3, 4, 5]

even_numbers = list(filter(lambda num: num % 2 == 0, numbers))

print(even_numbers)
```

### Combined example
- Here `filter()` first selects even numbers.
- Then `map()` converts those even numbers into squares.
- `lambda` is used to write both small functions.

```python
# Combined example of lambda, filter, and map
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda num: num % 2 == 0, numbers)
square_of_even_numbers = map(lambda num: num * num, even_numbers)

print(list(square_of_even_numbers))
```

### Final short difference
- Use `lambda` when you need a small function.
- Use `map()` when you want to change every item.
- Use `filter()` when you want to select only some items.
