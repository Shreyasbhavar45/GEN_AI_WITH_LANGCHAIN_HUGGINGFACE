## __MAP FUNCTION__

### Map function in simple language
- The `map()` function applies one function to every item in a list, tuple, or other iterable.
- It is used when we want to perform the same operation on many values.
- It takes two main things: a function and an iterable.
- It does not change the original list.
- It returns a map object, so we usually convert it into a list using `list()`.
- In simple words: `map()` means do this same work for every item.

### Syntax
```python
map(function, iterable)
```

````python
# Example without map function
numbers = [1, 2, 3, 4, 5]
squares = []

for num in numbers:
    squares.append(num * num)

print(squares)
````

````python
# Same example using map function
def square(num):
    return num * num

numbers = [1, 2, 3, 4, 5]
result = map(square, numbers)

print(list(result))
````

````python
# Map function with lambda function
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda num: num * num,numbers))

print(squares)
````

````python
# Another example: convert names to uppercase
names = ["shreyas", "rahul", "aman"]

upper_names = list(map(str.upper, names))

print(upper_names)
````

````python
#can Map have multiple iterables
number1 = [1,2,3]
number2 = [4,5,6]

added_numbers = list(map(lambda x,y:x+y,number1,number2))
print(added_numbers)
````

````python
#map convert a list of string to integers.
#use map to convert string to integer

str_numbers = ['1','2','3','4','5']
int_numbers = list(map(int,str_numbers))
print(int_numbers)
````
