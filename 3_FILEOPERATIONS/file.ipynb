# Binary Files

A binary file stores data as raw bytes instead of normal readable text. Images, audio files, videos, compressed files, database files, and executable files are common examples of binary files.

In Python, binary files are opened using binary modes:

- `rb` means read binary
- `wb` means write binary
- `ab` means append binary

Binary data is represented using the `bytes` type. Example: `b'ABC'` contains the byte values for `A`, `B`, and `C`.

```python
# Writing bytes to a binary file
data = b'Hello binary file!\n'

with open('example.bin', 'wb') as file:
    file.write(data)

print('Binary file created successfully')
```

```python
# Reading bytes from a binary file
with open('example.bin', 'rb') as file:
    content = file.read()

print(content)
print(type(content))
```

## Bytes and Hexadecimal

A byte is a number from `0` to `255`. Binary files are often inspected in hexadecimal because hex is easier to read than raw bits.

```python
# Inspecting byte values
data = b'ABC'

print(data[0])  # ASCII value of A
print(data[1])  # ASCII value of B
print(data[2])  # ASCII value of C

print(data.hex())
```

## Storing Numbers in Binary

If we store the number `1000` as text, Python writes characters like `'1'`, `'0'`, `'0'`, `'0'`. If we store it as binary, Python stores the actual numeric value using bytes.

The `struct` module helps convert Python values into binary data and convert binary data back into Python values.

```python
import struct

# '<i' means little-endian signed integer
number = 1000
binary_number = struct.pack('<i', number)

print(binary_number)
print(binary_number.hex())
```

```python
# Writing and reading a binary integer
with open('number.bin', 'wb') as file:
    file.write(binary_number)

with open('number.bin', 'rb') as file:
    stored_bytes = file.read()

restored_number = struct.unpack('<i', stored_bytes)[0]

print(stored_bytes)
print(restored_number)
```

## Text File vs Binary File

The same value can take different space depending on whether it is stored as text or binary.

```python
text_value = '1000'
binary_value = struct.pack('<i', 1000)

print('Text bytes:', text_value.encode('utf-8'))
print('Text size:', len(text_value.encode('utf-8')), 'bytes')

print('Binary bytes:', binary_value)
print('Binary size:', len(binary_value), 'bytes')
```

## Important Points

- Binary files are not meant to be read directly by humans.
- They are read by programs that understand the file format.
- Binary files are often smaller and faster for storing numeric or complex data.
- Opening a binary file in a text editor may show strange symbols.
- Small byte changes can corrupt binary files, especially images, videos, archives, and databases.

# Working with File Paths

A file path tells Python where a file or folder is located. File paths are important when reading files, writing files, creating folders, or checking whether a file exists.

There are two main types of paths:

- Absolute path: full path from the root of the system
- Relative path: path from the current working directory

```python
# Checking the current working directory
import os

print(os.getcwd())
```

## Relative Path

A relative path starts from the current working directory. If `example.txt` is in the same folder as the notebook, we can use only the file name.

```python
# Reading a file using a relative path
with open('example.txt', 'r') as file:
    content = file.read()

print(content)
```

## Absolute Path

An absolute path gives the complete location of a file. It is specific to your computer.

```python
# Getting the absolute path of a file
absolute_path = os.path.abspath('example.txt')

print(absolute_path)
```

## Using pathlib

`pathlib` is the modern and recommended way to work with file paths in Python. It makes paths easier to read and works well across operating systems.

```python
from pathlib import Path

path = Path('example.txt')

print(path)
print(path.exists())
print(path.is_file())
print(path.resolve())
```

```python
# Useful pathlib properties
path = Path('example.txt')

print('File name:', path.name)
print('File stem:', path.stem)
print('File extension:', path.suffix)
print('Parent folder:', path.parent)
```

## Joining Paths

Do not join paths manually using string concatenation. Use `/` with `Path` objects instead.

```python
folder = Path('data')
file_name = 'notes.txt'

file_path = folder / file_name

print(file_path)
```

## Creating Folders and Files

```python
# Create a folder and write a file inside it
folder = Path('data')
folder.mkdir(exist_ok=True)

file_path = folder / 'notes.txt'
file_path.write_text('This file was created using pathlib.')

print(file_path.read_text())
```

```python
# Create nested folders
nested_folder = Path('data/files/text')
nested_folder.mkdir(parents=True, exist_ok=True)

print(nested_folder.exists())
print(nested_folder.is_dir())
```

## Listing Files

```python
# List files and folders in the current directory
current_folder = Path('.')

for item in current_folder.iterdir():
    print(item)
```

```python
# Find only text files in the current folder
for file_path in Path('.').glob('*.txt'):
    print(file_path)
```

```python
# Find text files recursively inside folders
for file_path in Path('.').rglob('*.txt'):
    print(file_path)
```

## Important Points About File Paths

- A path tells Python where a file or folder is located.
- A relative path depends on the current working directory.
- An absolute path gives the full location of a file or folder.
- Use `Path` from `pathlib` for modern Python code.
- Use `path.exists()` before reading a file if you are not sure it exists.
- Use `folder / file_name` to join paths instead of string concatenation.
- Use `mkdir(exist_ok=True)` to avoid errors when a folder already exists.
- Use `mkdir(parents=True, exist_ok=True)` for nested folders.
- Use `glob()` to search in one folder and `rglob()` to search recursively.
- Avoid hardcoding absolute paths when sharing code with others.
