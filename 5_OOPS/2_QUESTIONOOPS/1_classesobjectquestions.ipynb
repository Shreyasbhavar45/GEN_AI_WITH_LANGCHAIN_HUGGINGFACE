### Assignment 4: Class with Private Attributes

Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Create an object of the class and perform some operations.

```python
class BankAccount:

    def __init__(self):
        self.__pin =''
        self.__account_number = ''
        self.__balance = 0
        self.menu()

    def menu(self):

        user_input = input("""
            "Hi How can I help You?"
            "0.Press 0 to write Account Number"
            "1.Press 1 to create a pin"
            "2.Press 2 to Change pin"
            "3.Press 3 to  Check Balance"
            "4.Press 4 to  Withdraw Money"
            "5.Press 5 to Deposit Money"
        """)

        if user_input == "0":
            #account number
            self.account_number()
    
        elif user_input == "1":
            # create Pin
            self.create_pin()
            
        elif user_input == "2":
            #change Pin
            self.change_pin()

        elif user_input == "3":
            #check Balance
            self.check_balance()

        elif user_input == "4":
            #withdraw Money
            self.withdraw()

        elif user_input == "5":
            #deposit Money
            self.deposit()
            
        else:
            exit

    def account_number(self):

        account_no = int(input("Enter Your Account Number"))
        self.__account_number = account_no

        print("Account Number Created Successfully")

        self.menu()
        
    def create_pin(self):

        user_pin = int(input("Enter Your Pin"))
        self.__pin = user_pin

        user_balance = int(input("Enter Your Balance"))
        self.__balance = user_balance

        print("Pin created Succesfully")

        self.menu()

    def change_pin(self):
        old_pin = (input('Enter old pin'))

        if old_pin == self.__pin:
            #Let him change the pin
            new_pin = input('Enter new Pin')
            self.__pin = new_pin
            print('pin change succesfully')
            self.menu()

        else:
            print('You cant change the pin')
            self.menu()

    def check_balance(self):

        user_pin = int(input('Enter Your Pin'))
        if user_pin == self.__pin:
            print('Your Balance is',self.__balance)

        else:
            print('You Enter Wrong Pin')

    def withdraw(self):

        user_pin = int(input('Enter the pin'))
        if user_pin == self.__pin:
            #Allowed to withdraw
            amount = int(input('Enter The amount'))

            if amount <= self.__balance:
                self.__balance = self.__balance - amount

                print('Withdrawl Succesful balance is',self.__balance)
            else:
                print('You have Unsufficient Balance')
        
        else:
            print("You Entered wrong pin")

        self.menu()




    
    
        
```
