## What is Aggregation?
- one class owns the other class

# example:
1 - Restraunt own a menu class

```python
#example 
class Customer:

    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address
    
    # we have to create a method to print the customer address
    def print_address(self):
        print("I am From",(self.address.get_city()),"My Pincode is",(self.address.pin),self.address.state)
    

class Address:

    def __init__(self,city,pin,state):
        self.__city = city
        self.pin = pin
        self.state = state

    # In aggregation You cannot fetch you own data means if you make private you cant access
    #setter for access or change the value in aggreagtion
    def get_city(self):
        return self.__city

# create a object of Class Address
add1 = Address("Nashik",422009,"Maharashtra")
# we want to create a object of customer class
cust = Customer("Shreyas","Male",add1)

#print the answer using by method calling
cust.print_address()
```

## __2.Inheritance__
- What is Inheritance
- example 
- what gets inherited ?

```python
# Inheritance And Its Benefits
```

```python
# Example

#parent class
class User:

    def __init__(self):
        self.name = "Shreyas"

    def login(self):
        print("Login")

# child class
class Student(User):

    def enroll(self):
        print("Enroll Into The Course")


u = User()
s = Student()

print(s.name)
s.login()
s.enroll()
```

```python
# What gets Inherited?
# - Constructor
# - Non Private Attributes
# - Non Private Methods
```

```python
# Constructor Exmple

class Phone:

    def __init__(self,price,brand,camera):
        print("Inside Phone Constructor")
        self.price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying A phone")

# child class
class SmartPhone(Phone):
    pass


s = SmartPhone(20000,"Apple",13)
```
