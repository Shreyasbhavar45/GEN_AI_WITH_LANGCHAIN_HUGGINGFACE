### __Reference Variables__

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object

```python
# Object without a refrence
class Person:

    def __init__(self):
        self.name = "Nitish"
        self.gender = "Male"

#p is the reference variable hold the objectss without p also we can access the objectt or create reference varibale 
p = Person()
q = p
```

```python
print(id(p))
print(id(q))
```

```python
print(p.name)
print(q.name)

q.name = "Ankit"
print(q.name)
print(p.name)
```

## __Pass By Reference__

```python
class Person:

    def __init__(self,name,gender):
        self.name = name 
        self.gender = gender


#Outside the Class -> Function
def greet(person):
    print("Hi my name is",person.name,'and I am a',person.gender)

    p1 = Person('Ankit','Male')
    return p1

p = Person("Nitish","Male")
x = greet(p)
print(x.name)
print(x.gender)
```

```python
# Creating a class named Person
class Person:

    # Constructor
    # It runs automatically when we create a Person object
    def __init__(self, name, gender):

        # Store the given name inside the object
        self.name = name

        # Store the given gender inside the object
        self.gender = gender


# Function outside the class
# 'person' is a parameter that will receive a Person object
def greet(person):

    # Print the ID of the object received by the function
    print(id(person))

    # Access the object's name and gender using the person parameter
    print("Hi my name is", person.name, "and I am a", person.gender)

    # Create another Person object
    p1 = Person("Ankit", "Male")

    # Return the newly created p1 object
    return p1


# Create a Person object
# p is the reference to this object
p = Person("Nitish", "Male")

# Print the ID of the object referenced by p
print(id(p))

# Pass the object p to the greet() function
# 'person' inside greet() now refers to the same object as p
greet(p)
```

## __WhaT is Encapsulation__

```python
# Creating A Atm Class

class Atm:

    # Constructor :- A function inside the class
    # It is a special Function it have superpower
    # Superpower of constructor is automatically call we dont need to call consturctor
    def __init__(self):
        self.pin = ' '
        self.balance = 0
        self.menu()
    
    def menu(self):
        user_input = input("""
                           
        Hi How can I help You ?
        1.Press 1 to create pin
        2.Press 2 to change pin
        3.Press 3 to check balance
        4.Press 4 to with draw
        5.Anything else to exit
""")
        
        if user_input == '1':
            #Create pin
            self.create_pin()

        elif user_input == '2':
            #change pin
            self.change_pin()
        elif user_input == '3':
            #check Balance
            self.check_balance()
        elif user_input == '4':
            #withdraw
            self.withdraw()
        else:
            exit
 
    def create_pin(self):

        user_pin = input('Enter Your Pin')
        self.pin = user_pin

        user_balance = int(input("Enter The Balance"))
        self.balance = user_balance

        print('Pin Created Succesfully')
        self.menu()

    def change_pin(self):
        old_pin = input('Enter old pin')

        if old_pin == self.pin:
            #Let him change the pin
            new_pin = input('Enter new Pin')
            self.pin = new_pin
            print('pin change succesfully')
            self.menu()

        else:
            print('You cant change the pin')
            self.menu()

    def check_balance(self):

        user_pin = input('Enter Your Pin')
        if user_pin == self.pin:
            print('Your Balance is',self.balance)

        else:
            print('You Enter Wrong Pin')

    def withdraw(self):

        user_pin = input('Enter the pin')
        if user_pin == self.pin:
            #Allowed to withdraw
            amount = int(input('Enter The amount'))

            if amount <= self.balance:
                self.balance = self.balance - amount

                print('Withdrawl Succesful balance is',self.balance)
            else:
                print('You have Unsufficient Balance')
        
        else:
            print("You Entered wrong pin")

        self.menu()
```
