```python
#3_Your First Class
class Student:
    pass

# class is the keyword that says "I'm making a blueprint".
# Student is the name. Convention: start with a capital letter, and use CamelCase for longer names (BankAccount, SmartPhone).
# : and the indented body work just like if or def.
# pass means "empty for now, do nothing". It's needed because a body can't be blank.
```

```python
##4_ Creating Objects
s1 = Student()
s2 = Student()

# Student() (with brackets) means "build a new object from this blueprint". The object is stored in the variable s1.
```

```python
print(s1)
print(type(s1))
```

```python
# Important: s1 and s2 are two different objects, even though they came from the same class.
print(s1 == s2)#False
print(s1 is s2)#False
print(id(s1),id(s2))#Two Diffrent numbers
```

```python
#5_Giving Objects data (attributes)
# Our class is empty, but Python lets us attach data to an object directly:
s1.name = 'Riya'
s1.marks = 85

s2.name = 'Aman'
s2.marks = 72

print(s1.name,s1.marks)
print(s2.name,s2.marks)


# The dot . means "go inside this object".
# name and marks are called attributes (data stored inside an object).
# # Each object keeps its own copy. Changing s1.marks doesn't touch s2.marks.
```

```python
s1.marks = 95 # change
print(s1.marks) #95
print(s2.marks)  #72 (unchanged)
```

```python
# Trying to read something that doesn't exist:
print(s1.age)
```

```python
#6.Handy built-in tools
print(isinstance(s1,Student)) #True : 'is s1 an object of student?'
print(hasattr(s1,"name"))     #True: "does s1 have an attribute name?"
print(getattr(s1,"name"))     # Riya
setattr(s1, "city", "Nashik")    # same as s1.city = "Nashik"
print(s1.__dict__)               # {'name': 'Riya', 'marks': 95, 'city': 'Nashik'}

# __dict__ is a peek at everything stored inside the object, which is very useful for debugging.
```

```python
# 7. The problem with what we did

# Setting attributes one by one after creating the object is tedious, 
# and we can easily forget one (s2 might never get marks). In Lesson 2 we fix this with __init__, 
# which sets up every object properly at the moment it is created.
```

```python
# Creating A Atm Class

class Atm:

    # Constructor :- A function inside the class
    # It is a special Function it have superpower
    # Superpower of constructor is automatically call we dont need to call consturctor
    def __init__(self):
        self.pin = ' '
        self.balance = 0
        self.menu()
    
    def menu(self):
        user_input = input("""
                           
        Hi How can I help You ?
        1.Press 1 to create pin
        2.Press 2 to change pin
        3.Press 3 to check balance
        4.Press 4 to with draw
        5.Anything else to exit
""")
        
        if user_input == '1':
            #Create pin
            self.create_pin()

        elif user_input == '2':
            #change pin
            self.change_pin()
        elif user_input == '3':
            #check Balance
            self.check_balance()
        elif user_input == '4':
            #withdraw
            self.withdraw()
        else:
            exit
 
    def create_pin(self):

        user_pin = input('Enter Your Pin')
        self.pin = user_pin

        user_balance = int(input("Enter The Balance"))
        self.balance = user_balance

        print('Pin Created Succesfully')
        self.menu()

    def change_pin(self):
        old_pin = input('Enter old pin')

        if old_pin == self.pin:
            #Let him change the pin
            new_pin = input('Enter new Pin')
            self.pin = new_pin
            print('pin change succesfully')
            self.menu()

        else:
            print('You cant change the pin')
            self.menu()

    def check_balance(self):

        user_pin = input('Enter Your Pin')
        if user_pin == self.pin:
            print('Your Balance is',self.balance)

        else:
            print('You Enter Wrong Pin')

    def withdraw(self):

        user_pin = input('Enter the pin')
        if user_pin == self.pin:
            #Allowed to withdraw
            amount = int(input('Enter The amount'))

            if amount <= self.balance:
                self.balance = self.balance - amount

                print('Withdrawl Succesful balance is',self.balance)
            else:
                print('You have Unsufficient Balance')
        
        else:
            print("You Entered wrong pin")

        self.menu()
```
