# Exception Handling in Python

Exception handling is used to handle runtime errors safely. Instead of stopping the whole program when an error occurs, Python allows us to catch the error and decide what should happen next.

The main blocks are:

- `try`: write the code that may cause an error.
- `except`: handle the error if it happens.
- `else`: run code only when no error happens in the `try` block.
- `finally`: run code every time, whether an error happens or not.

## Basic Syntax

```python
try:
    # risky code
except SomeError:
    # error handling code
else:
    # runs only if try has no error
finally:
    # always runs
```

Think of it like this:

- Python first enters the `try` block.
- If an error occurs, Python jumps to a matching `except` block.
- If no error occurs, Python runs the `else` block.
- At the end, Python always runs the `finally` block.

## 1. `try` Block

The `try` block contains code that may raise an exception.

Example: dividing a number by zero causes a `ZeroDivisionError`.

````python
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print("Error: You cannot divide by zero.")
````

### Explanation

- `10 / 0` is risky code, so it is placed inside `try`.
- Python gets a `ZeroDivisionError`.
- The `except ZeroDivisionError` block handles the error.
- The program does not crash.

## 2. `except` Block

The `except` block catches and handles exceptions. You can handle different error types separately.

This is better than using one general `except` for every error because each error may need a different solution.

````python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print("Result:", result)
except ValueError:
    print("Error: Please enter only a valid number.")
except ZeroDivisionError:
    print("Error: Number cannot be zero.")
````

### Explanation

- If the user enters text like `abc`, `int()` raises `ValueError`.
- If the user enters `0`, division raises `ZeroDivisionError`.
- Each error is handled by its own `except` block.

## 3. Catching the Exception Object

You can store the actual exception message in a variable using `as`.

````python
try:
    marks = int("ninety")
except ValueError as error:
    print("Something went wrong:", error)
````

### Explanation

- `error` stores the original error message.
- This is useful for debugging and logging.

## 4. `else` Block

The `else` block runs only when the `try` block completes successfully.

Use `else` for code that should run only after the risky operation succeeds.

````python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input. Please enter a number.")
else:
    square = number ** 2
    print("Square of the number is:", square)
````

### Explanation

- If input conversion fails, `except` runs and `else` is skipped.
- If input conversion succeeds, `else` runs.
- This keeps success logic separate from error handling logic.

## 5. `finally` Block

The `finally` block always runs.

It is commonly used for cleanup tasks such as closing files, closing database connections, or releasing resources.

````python
try:
    number = int(input("Enter a number: "))
    result = 50 / number
except ValueError:
    print("Invalid input. Please enter a number.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Division successful. Result:", result)
finally:
    print("This finally block always runs.")
````

### Explanation

Try different inputs:

- Enter `10`: `try`, `else`, and `finally` run.
- Enter `0`: `try`, matching `except`, and `finally` run.
- Enter `hello`: `try`, matching `except`, and `finally` run.

`finally` runs in all cases.

## Complete Example: File Handling with `try`, `except`, `else`, and `finally`

This example shows a practical use case. We try to open and read a file. If the file is found, we read it. If not, we handle the error. Finally, we close the file if it was opened.

````python
file = None

try:
    file = open("sample.txt", "r")
except FileNotFoundError:
    print("Error: The file was not found.")
else:
    content = file.read()
    print("File content:")
    print(content)
finally:
    if file is not None:
        file.close()
        print("File closed successfully.")
    else:
        print("No file was opened, so nothing to close.")
````

### Explanation

- `file = None` is used before `try` so we can safely check it later.
- `open("sample.txt", "r")` may raise `FileNotFoundError`.
- If the file does not exist, `except` handles the error.
- If the file opens successfully, `else` reads the file.
- `finally` checks whether the file was opened and closes it safely.

## Flow of Execution

| Situation | Blocks That Run |
|---|---|
| No error in `try` | `try` -> `else` -> `finally` |
| Error in `try` and matching `except` exists | `try` -> `except` -> `finally` |
| Error in `try` and no matching `except` exists | `try` -> `finally` -> program shows error |

Important point: `else` does not run if an exception happens.

## Best Practices

- Catch specific exceptions like `ValueError`, `ZeroDivisionError`, or `FileNotFoundError`.
- Avoid writing only `except:` because it catches every error, including errors you may not expect.
- Keep only risky code inside the `try` block.
- Use `else` for code that should run only after success.
- Use `finally` for cleanup code.
- Do not silently ignore errors unless you have a clear reason.

## Small Practice Program

This program asks for two numbers and divides them safely.

````python
try:
    first_number = float(input("Enter first number: "))
    second_number = float(input("Enter second number: "))
    answer = first_number / second_number
except ValueError:
    print("Please enter valid numeric values.")
except ZeroDivisionError:
    print("The second number cannot be zero.")
else:
    print("Division completed successfully.")
    print("Answer:", answer)
finally:
    print("Program execution finished.")
````
